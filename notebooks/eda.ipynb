# Khám phá dữ liệu (EDA) - Tập dữ liệu SMS Spam Collection

Notebook này phân tích các đặc trưng thống kê của tập dữ liệu tin nhắn SMS rác (Spam) và thường (Ham).

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

# Thiết lập giao diện đồ thị
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

## 1. Tải và xem cấu trúc dữ liệu

```python
data_path = "../data/raw/SMSSpamCollection"
if os.path.exists(data_path):
    df = pd.read_csv(data_path, sep='\t', names=['label', 'text'], header=None)
    print(f"Kích thước dữ liệu: {df.shape}")
    display(df.head())
else:
    print("Chưa tải dữ liệu. Vui lòng chạy lệnh: python ../src/download_data.py trước.")
```

## 2. Phân phối của Nhãn (Label Distribution)

```python
if 'df' in locals():
    label_counts = df['label'].value_counts()
    print(label_counts)
    
    # Biểu đồ cột phân phối lớp
    plt.figure(figsize=(6, 4))
    sns.countplot(x='label', data=df, hue='label', palette='Set2', legend=False)
    plt.title('Phân phối nhãn Tin nhắn (Ham vs Spam)')
    plt.ylabel('Số lượng')
    plt.xlabel('Nhãn')
    plt.show()
```

## 3. Độ dài tin nhắn (Message Length)

```python
if 'df' in locals():
    df['length'] = df['text'].apply(len)
    
    # Biểu đồ phân phối độ dài
    plt.figure(figsize=(10, 5))
    sns.histplot(data=df, x='length', hue='label', bins=50, kde=True, palette='Set1', multiple='stack')
    plt.title('Phân phối độ dài tin nhắn (Số ký tự)')
    plt.xlabel('Độ dài tin nhắn')
    plt.ylabel('Tần suất')
    plt.xlim(0, 300) # Giới hạn hiển thị để dễ nhìn
    plt.show()
```

Thống kê mô tả về độ dài tin nhắn của từng loại:

```python
if 'df' in locals():
    print(df.groupby('label')['length'].describe())
```

## 4. Các từ phổ biến nhất trong Spam vs Ham

```python
import sys
sys.path.append('../src')
from preprocessing import clean_text

if 'df' in locals():
    df['clean_text'] = df['text'].apply(clean_text)
    
    spam_words = ' '.join(df[df['label'] == 'spam']['clean_text']).split()
    ham_words = ' '.join(df[df['label'] == 'ham']['clean_text']).split()
    
    spam_counter = Counter(spam_words).most_common(20)
    ham_counter = Counter(ham_words).most_common(20)
    
    # Vẽ biểu đồ các từ phổ biến nhất trong Spam
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    words, counts = zip(*spam_counter)
    sns.barplot(x=list(counts), y=list(words), palette='Reds_r')
    plt.title('Top 20 từ xuất hiện nhiều nhất trong SPAM')
    plt.xlabel('Số lần xuất hiện')
    
    # Vẽ biểu đồ các từ phổ biến nhất trong Ham
    plt.subplot(1, 2, 2)
    words, counts = zip(*ham_counter)
    sns.barplot(x=list(counts), y=list(words), palette='Blues_r')
    plt.title('Top 20 từ xuất hiện nhiều nhất trong HAM')
    plt.xlabel('Số lần xuất hiện')
    
    plt.tight_layout()
    plt.show()
```
